```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
image = cv2.imread('pexels-alexravvas-26611787.jpg')
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
image1 = image.copy()
gray = cv2.cvtColor(image,cv2.COLOR_BGR2GRAY)
```

```python
edges = cv2.Canny(gray,50, 150, apertureSize =3)
lines = cv2.HoughLines(edges, 1, np.pi/180 , 50)
print("Total Number of Lines = " , len(lines))
print(lines)
for rho,theta in lines[0]:
    a = np.cos(theta)
    b = np.sin(theta)
    x0 = a*rho
    y0 = b*rho
    x1 = int(x0+1000*(-b))
    y1 = int(y0+1000*(a))
    x2 = int(x0-1000*(-b))
    y2 = int(y0-1000*(a))
    cv2.line(image1,(x1,y1), (x2,y2),(0,0,255),2)
    
plt.figure()
plt.subplot(121)
plt.title(" Edges")
plt.imshow(edges,cmap = 'gray')
plt.subplot(122)
plt.title("Lines By Hough Method")
plt.imshow(image1) 
plt.tight_layout()   
plt.show()
```
