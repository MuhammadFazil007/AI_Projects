### Importing all Required Images

```python
import numpy as np
import cv2
import matplotlib.pyplot as plt
```

### Reading the Image

```python
image = cv2.imread('pexels-alexravvas-26611787.jpg')
plt.figure()
plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
plt.title(' Image Readed Through Open CV (CV2)')
plt.show()
# cv2.imshow('original', image)
# cv2.waitKey(0)
# cv2.destroyAllWindows()
```

### Converting to Gray Scale

```python
resized_image = cv2.resize(image , (500,500))
image2 = cv2.cvtColor(resized_image,cv2.COLOR_BGR2GRAY)
plt.figure()
plt.imshow(image2)
plt.title('Gray Image ')
plt.show()
```

```python
negative_image1 = cv2.bitwise_not(image2)
negative_image2 = 255 - image2
```

```python
# Subplots 
plt.figure(figsize=(20,20))
plt.subplot(1,4,1)
plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
plt.title("Original Image")
plt.subplot(1,4,2)
plt.imshow(cv2.cvtColor(image2, cv2.COLOR_BGR2RGB))
plt.title("Gray Scale Image")
plt.subplot(1,4,3)
plt.imshow(negative_image1, cmap='gray')
plt.title(" Negative Image using Bitwise NOT")
plt.subplot(1,4,4)
plt.imshow(negative_image2 , cmap = 'gray')
plt.title(" Negative Image using 255 - pixel value")
plt.tight_layout()
plt.show()




```

### Log Tranformed Image

```python
c = 255 / np.log(1 + np.max(image))
log_transformed = c*np.log(1 + image)
log_image = np.array(log_transformed, dtype = np.uint8)
```

```python
plt.figure()
plt.subplot(1,2,1)
plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
plt.title(" Original Image")
plt.subplot(1,2,2)
plt.imshow(log_image)
plt.title("Log Transformed Image")
plt.tight_layout()
plt.show()
```

```python
gamma = 2
antilog_image = np.exp(np.power(image2 , gamma) / c) - 1
plt.imshow(antilog_image,cmap = 'gray')
```
