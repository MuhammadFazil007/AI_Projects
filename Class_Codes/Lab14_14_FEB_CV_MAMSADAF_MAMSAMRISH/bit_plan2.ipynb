### Importing all Required Libraries

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import cv2
```

### Reading the Image

```python
# Reading a picture in grascale

image = cv2.imread('pexels-alexravvas-26611787.jpg', cv2.IMREAD_GRAYSCALE)
image2 = cv2.imread('pexels-alexravvas-26611787.jpg')
cv2.imshow('original', image)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

### Displaying the Image Using Matplotlib

```python
plt.figure(figsize = (10,10))
plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
plt.title('Original Image')
plt.show()
```

### Printing Shape of the Image

```python
print(f" Shape of the Image:: {image.shape}")
print(f" Type of the Image:: {type(image)}")
```

### Resizing the Image & Reshaping the Image

```python
image1 = image.copy()
image3 = cv2.resize(image1,(500,500))
plt.figure(figsize = (10,10))
plt.imshow(cv2.cvtColor(image2, cv2.COLOR_BGR2RGB))
plt.title('Image 2')
plt.show()
```

```python
thresh , image4 = cv2.threshold(image3 , 150 , 255 , cv2.THRESH_BINARY)
plt.figure(figsize = (15,10))
plt.subplot(121)
plt.imshow(image3 , cmap = 'gray')
plt.title("Original Image")
plt.subplot(122)
plt.imshow(image4 , cmap = 'gray')
plt.title(" Thresholded Image")
```

```python
thresh1 , threshimg1 = cv2.threshold(image3 , 127 , 255 , cv2.THRESH_BINARY)
thresh2 , threshimg2 = cv2.threshold(image3 , 127 , 255 , cv2.THRESH_BINARY_INV)
thresh3 , threshimg3 = cv2.threshold(image3 , 127 , 255 , cv2.THRESH_TRUNC)
thresh4 , threshimg4 = cv2.threshold(image3 , 127 , 255 , cv2.THRESH_TOZERO)
thresh5 , threshimg5 = cv2.threshold(image3 , 127 , 255 , cv2.THRESH_TOZERO_INV)

# (Thresh Binary)=> Pixels greater than 127 converted to 255
# (Thresh Binary)=> Pixesls less than 127 converted to 0
# (Thresh Binary INV) => Pixels greater than 127 converted to 0
#  
```

```python
titles = ['Original' , 'Binary' , 'Thresh_BINARY_INV' , 'THRESH_TRUNC' , 'THRESH_TOZERO' , 'THRESH_TROZERO_INV']
images = [image3 , threshimg1 , threshimg2 , threshimg3 , threshimg4, threshimg5]
for i in range(6):
    plt.subplot(2,3,i+1)
    plt.imshow(images[i] , cmap = 'gray', vmin = 0 , vmax = 255)
    plt.title(titles[i])
    plt.xticks([]),plt.yticks([])

plt.show()
```

```python
image5 = cv2.imread('pexels-alexravvas-26611787.jpg',cv2.IMREAD_GRAYSCALE)
# image5 = cv2.cvtColor(image5 , cv2.COLOR_BGR2RGB)
mask33 = np.ones((3,3) , dtype = np.float32) / (3*3)
mask55 = np.ones((5,5) , dtype = np.float32) / (5*5)
mask99 = np.ones((9,9) , dtype = np.float32) / (9*9)
mask1313 = np.ones((13,13) , dtype = np.float32) / (13*13)
mask1717 = np.ones((17,17) , dtype = np.float32) / (17*17)

```

```python
filtered33 = cv2.filter2D(image5 , -1 , mask33)
filtered55 = cv2.filter2D(image5 , -1 , mask55)
filtered99 = cv2.filter2D(image5 , -1 , mask99)
filtered1313 = cv2.filter2D(image5 , -1 , mask1313)
filtered1717 = cv2.filter2D(image5 , -1 , mask1717)
```

```python
plt.figure(figsize = (15,10))
plt.subplot(161)
plt.imshow(image5 , cmap = 'gray')
plt.title("Original Image")
plt.subplot(162)
plt.imshow(filtered33 , cmap = 'gray')
plt.title(" 3x3 Filtered Image")
plt.subplot(163)
plt.imshow(filtered55 , cmap = 'gray')
plt.title(" 5x5 Filtered Image")
plt.subplot(164)
plt.imshow(filtered99 , cmap = 'gray')
plt.title(" 9x9 Filtered Image")
plt.subplot(165)
plt.imshow(filtered1313 , cmap = 'gray')
plt.title(" 13x13 Filtered Image")
plt.subplot(166)
plt.imshow(filtered1717 , cmap = 'gray')
plt.title(" 17x17 Filtered Image")
plt.tight_layout()
plt.show()
```
