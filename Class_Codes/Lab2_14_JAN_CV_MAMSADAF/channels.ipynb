### Importing all Required Libraries

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import cv2
```

### Reading the Image

```python
image = cv2.imread('car_gray_wet_147750_1366x768.jpg')
cv2.imshow('original', image)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

### Displaying the Image Using Matplotlib

```python
plt.figure(figsize = (10,10))
plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
plt.title('Original Image')
plt.show()
```

### Printing Shape of the Image

```python
print(f" Shape of the Image:: {image.shape}")
print(f" Type of the Image:: {type(image)}")
```

### Resizing the Image & Reshaping the Image

```python
image1 = image.copy()
image2 = cv2.resize(image1,(500,500))
plt.figure(figsize = (10,10))
plt.imshow(cv2.cvtColor(image2, cv2.COLOR_BGR2RGB))
plt.title('Image 2')
plt.show()
```

### Separating Channels

```python
red = image2[0:500,0:500,0]
green = image2[:,:,1]
blue = image2[0:500,0:500,2]
```

```python
plt.figure(figsize = (10,10))
plt.subplot(1,4,1)
plt.imshow(cv2.cvtColor(image2, cv2.COLOR_BGR2RGB))
plt.title('ALL Channels')
plt.subplot(1,4,2)
plt.imshow(red)
plt.subplot(1,4,3)
plt.imshow(green)
plt.subplot(1,4,4)
plt.imshow(blue)
plt.tight_layout()
plt.show()
```

```python
image3 = cv2.hconcat([red,green,blue])
plt.imshow(image3)
```
