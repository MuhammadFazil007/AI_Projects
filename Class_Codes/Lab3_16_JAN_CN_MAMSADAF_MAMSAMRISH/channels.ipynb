### Importing all Required Libraries

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import cv2
```

### Reading the Image

```python
image = cv2.imread('pexels-alexravvas-26611787.jpg')
# cv2.imshow('original', image)
# cv2.waitKey(0)
# cv2.destroyAllWindows()
```

### Displaying the Image Using Matplotlib

```python
plt.figure(figsize = (10,10))
plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
plt.title('Original Image')
plt.show()
```

### Printing Shape of the Image

```python
print(f" Shape of the Image:: {image.shape}")
print(f" Type of the Image:: {type(image)}")
```

### Resizing the Image & Reshaping the Image

```python
image1 = image.copy()
image2 = cv2.resize(image1,(500,500))
plt.figure(figsize = (10,10))
plt.imshow(cv2.cvtColor(image2, cv2.COLOR_BGR2RGB))
plt.title('Image 2')
plt.show()
```

### Separating Channels

```python
red = image2[0:500,0:500,0]
green = image2[:,:,1]
blue = image2[0:500,0:500,2]
```

```python
plt.figure(figsize = (10,10))
plt.subplot(1,4,1)
plt.imshow(cv2.cvtColor(image2, cv2.COLOR_BGR2RGB))
plt.title('ALL Channels')
plt.subplot(1,4,2)
plt.imshow(red)
plt.subplot(1,4,3)
plt.imshow(green)
plt.subplot(1,4,4)
plt.imshow(blue)
plt.tight_layout()
plt.show()
```

```python
image3 = cv2.vconcat([red,green,blue])
plt.imshow(image3)
```

```python
image3 = cv2.hconcat([red,green,blue])
plt.imshow(image3)
```

```python
image6 = cv2.cvtColor(image2, cv2.COLOR_BGR2RGB)
image7 = cv2.cvtColor(image2,cv2.COLOR_BGR2GRAY)
plt.figure()
plt.subplot(1,3,1)
plt.imshow(image2)
plt.title('Original')
plt.subplot(1,3,2)
plt.imshow(image6)
plt.title('RGB')
plt.subplot(1,3,3)
plt.imshow(image7,cmap='gray')
plt.title('Gray')
plt.show()
```

```python
image8 = image2.copy()
image8[:,:,0] = 0  # Red Array Is Set to 0
image8[:,:,1] = 0  # Green Array Is Set to 0
```

```python
image9 = image2.copy()
image9[:,:,0] = 0  # Red Array Is Set to 0
image9[:,:,2] = 0  # Blue Array Is Set to 0
```

```python
image10 = image2.copy()
image10[:,:,1] = 0  # Green Array Is Set to 0
image10[:,:,2] = 0  # Blue Array Is Set to 0
```

### Picture Coloring

```python
plt.figure(figsize = (10,10))
plt.subplot(1,4,1)
plt.imshow(image8)
plt.title('Blue Image')
plt.subplot(1,4,2)
plt.imshow(image9)
plt.title('Green Image')
plt.subplot(1,4,3)
plt.imshow(image10)
plt.title('Red Image')
plt.subplot(1,4,4)
plt.imshow(image2)
plt.title("Original Image")
plt.tight_layout()
plt.show()
```

### Picture Indexing

```python
rows = 250
cols = 300
plt.figure(figsize = (10,10))
plt.imshow(image2[100:rows,150:cols,:])
plt.title('Cropped Image')
plt.show()
```
