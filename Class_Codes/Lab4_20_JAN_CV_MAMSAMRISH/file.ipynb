```python
from PIL import Image
import matplotlib.pyplot as plt
import numpy as np
import cv2
```

### Reading & Displaying the Image

```python
image = Image.open('pexels-alexravvas-26611787.jpg')
image3 = cv2.imread('pexels-alexravvas-26611787.jpg')
image.show()
```

### Resizing the Image

```python
# Resizing the Image Using Pillow
resized_image = image.resize((800, 600))
resized_image.show()
```

### Plotting Multiple Images on 1 Graph

```python
#Plotting Multiple Images on 1 Graph
fig, axs = plt.subplots(1, 2)
axs[0].imshow(image)
axs[0].set_title('Original Image')
axs[1].imshow(resized_image)
axs[1].set_title('Resized Image')
plt.tight_layout()
for ax in axs:
    ax.axis('on')
plt.show()


```

### Seperating Channels of a colored Image

```python
# Seperating Channels of a colored Image
r, g, b = image.split()
fig, axs = plt.subplots(1, 3, figsize=(15, 5))
axs[0].imshow(r, cmap='Reds')
axs[0].set_title('Red Channel')
axs[1].imshow(g, cmap='Greens')
axs[1].set_title('Green Channel')
axs[2].imshow(b, cmap='Blues')
axs[2].set_title('Blue Channel')
plt.tight_layout()
for ax in axs:
    ax.axis('on')
plt.show()
```

### Converting RGB to GRB

```python
grb_image = Image.merge("RGB", (g, r, b))
grb_image.show()
```

### Horizontal Concatenation of Image

```python
horizontal_concat = Image.new('RGB', (image.width * 2, image.height))
horizontal_concat.paste(image, (0, 0))
horizontal_concat.paste(resized_image, (image.width, 0))
horizontal_concat.show()
```

### Vertical Concatenation of Image

```python
vertical_concat = Image.new('RGB', (image.width, image.height * 2))
vertical_concat.paste(image, (0, 0))
vertical_concat.paste(resized_image, (0, image.height))
vertical_concat.show()
```

### Flipping

```python
image4 = image3.copy()
width,height,channels = image4.shape
fipped_array = np.zeros((width,height,channels),dtype = np.uint8)
for i,row in enumerate(image4):
    fipped_array[width-1-i,:,:] = row

plt.figure()
plt.imshow(fipped_array)
plt.title('Flipping Along Y-Axis')
plt.show()
```

```python
image4 = cv2.flip(image4,1)
plt.figure()
plt.imshow(image4)
```
