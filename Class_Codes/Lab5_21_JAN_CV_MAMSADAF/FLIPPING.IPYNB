#### Importing all Required Libraries

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import cv2
from PIL import Image
```

### Reading the Image

```python
image = cv2.imread("pexels-gabriele-brancati-32566116-7627322.jpg")
cv2.imshow('original', image)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

### Displaying the Image using Matplotlib

```python
plt.figure(figsize = (10,10))
plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
plt.title('Original Image')
plt.show()
```

### Resizing the Image

```python
image1 = image.copy()
image2 = cv2.resize(image1,(500,500))
plt.figure()
plt.imshow(cv2.cvtColor(image2, cv2.COLOR_BGR2RGB))
plt.title('Image 2')
plt.show()
```

```python
width,height,channels = image2.shape
print(image2.shape)
```

### Flipping Along Y-Axis

```python
width,height,channels = image2.shape
fipped_array = np.zeros((width,height,channels),dtype = np.uint8)
for i,row in enumerate(image2):
    fipped_array[width-1-i,:,:] = row

plt.figure()
plt.imshow(fipped_array)
plt.title('Flipping Along Y-Axis')
plt.show()
```

### Flipping Along X-Axis

```python
width,height,channels = image2.shape
fipped_array = np.zeros((width,height,channels),dtype = np.uint8)
for y in range(height):
    for x in range(width):
        fipped_array[y,x] = image2[y,width - x - 1]
plt.figure()
plt.imshow(fipped_array)
plt.title('Flipping Along X-Axis')
plt.show()
```

```python
image4 = cv2.flip(image2,1)
for flipcode in [0,1,-1]:
    flip_array = cv2.flip(image2, flipcode)
    plt.imshow(flip_array)
    plt.title("FlipCode"+str(flipcode))
    plt.show()

plt.figure(figsize = (10,10))
plt.subplot(1,4,1)
plt.imshow(cv2.cvtColor(image2, cv2.COLOR_BGR2RGB))
plt.title('ALL Channels')
plt.subplot(1,4,2)
plt.imshow(red)
plt.subplot(1,4,3)
plt.imshow(green)
plt.subplot(1,4,4)
plt.imshow(blue)
plt.tight_layout()
plt.show()    
```
