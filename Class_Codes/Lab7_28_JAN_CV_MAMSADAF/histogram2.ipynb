### Importing all Required Libraries

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import cv2
```

### Reading the Image

```python
image = cv2.imread('pexels-alexravvas-26611787.jpg')
cv2.imshow('original', image)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

### Displaying the Image Using Matplotlib

```python
plt.figure(figsize = (10,10))
plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
plt.title('Original Image')
plt.show()
```

### Printing Shape of the Image

```python
print(f" Shape of the Image:: {image.shape}")
print(f" Type of the Image:: {type(image)}")
```

### Resizing the Image & Reshaping the Image

```python
image1 = image.copy()
image2 = cv2.resize(image1,(500,500))
plt.figure(figsize = (10,10))
plt.imshow(cv2.cvtColor(image2, cv2.COLOR_BGR2RGB))
plt.title('Image 2')
plt.show()
```

### Separating Channels

```python
red = image2[0:500,0:500,0]
green = image2[:,:,1]
blue = image2[0:500,0:500,2]
```

```python
plt.figure(figsize = (10,10))
plt.subplot(1,4,1)
plt.imshow(cv2.cvtColor(image2, cv2.COLOR_BGR2RGB))
plt.title('ALL Channels')
plt.subplot(1,4,2)
plt.imshow(red)
plt.subplot(1,4,3)
plt.imshow(green)
plt.subplot(1,4,4)
plt.imshow(blue)
plt.tight_layout()
plt.show()
```

```python
image3 = cv2.vconcat([red,green,blue])
plt.imshow(image3)
```

```python
image3 = cv2.hconcat([red,green,blue])
plt.imshow(image3)
```

```python
image6 = cv2.cvtColor(image2, cv2.COLOR_BGR2RGB)
image7 = cv2.cvtColor(image2,cv2.COLOR_BGR2GRAY)
plt.figure()
plt.subplot(1,3,1)
plt.imshow(image2)
plt.title('Original')
plt.subplot(1,3,2)
plt.imshow(image6)
plt.title('RGB')
plt.subplot(1,3,3)
plt.imshow(image7,cmap='gray')
plt.title('Gray')
plt.show()
```

```python
image_colored = cv2.imread('pexels-alexravvas-26611787.jpg')
image_colored = cv2.resize(image_colored, (500,500))
image_colored = cv2.cvtColor(image_colored, cv2.COLOR_BGR2RGB)
image_grey = cv2.imread('pexels-alexravvas-26611787.jpg', cv2.IMREAD_GRAYSCALE)
image_grey = cv2.resize(image_grey , (500,500))
print(" Max Intensity in Colored " , image_colored.max())
print(" Min Intensity in Colored " , image_colored.min())
print(" Max Intensity in Grey " , image_grey.max())
print(" Min Intensity in Grey " , image_grey.min())
```

```python
r = np.zeros((256,)) # Size of one dimensional Zeros Array is 256
height, width = image_grey.shape
for i in range(height):
    for j in range(width):
        r[image_grey[i,j]] = r[image_grey[i,j]] + 1


plt.figure()
plt.bar(range(256), r )
plt.xlim(0, 256)
plt.xlabel(" Pixel Intensity")
plt.ylabel(" Frequency")
plt.title(" Histogram ")
plt.tight_layout()        
```

```python
hist = cv2.calcHist(images = [image_grey], channels = [0],mask = None, histSize= [256], ranges =[0,256])
plt.figure()
plt.bar(range(256), hist.ravel(), width = 1, color = "grey" )
plt.xlim(0, 256)
plt.xlabel(" Pixel Intensity")
plt.ylabel(" Frequency")
plt.title(" Histogram (GREY CHANNEL) ")
plt.tight_layout()        

```
