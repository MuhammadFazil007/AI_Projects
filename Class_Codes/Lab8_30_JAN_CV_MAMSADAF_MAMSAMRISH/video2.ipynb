# VIDEO PROCESSING AND ANALYSIS

### Importing all Required Libraries

```python
import cv2
import numpy as np
import os
from datetime import timedelta
```

## ============ Loading the Video =================

```python
video_path = 'Screen Recording 2025-12-17 022407.mp4'
video_process = cv2.VideoCapture(video_path)

if not video_process.isOpened():
    print(f"✗ Unable to open video: {video_path}")
    exit()

print(f"✓ Video loaded successfully: {video_path}\n")
```

## ==================== 1. GET VIDEO PROPERTIES ====================

```python
fps = video_process.get(cv2.CAP_PROP_FPS)
total_frames = int(video_process.get(cv2.CAP_PROP_FRAME_COUNT))
frame_width = int(video_process.get(cv2.CAP_PROP_FRAME_WIDTH))
frame_height = int(video_process.get(cv2.CAP_PROP_FRAME_HEIGHT))
duration_seconds = total_frames / fps if fps > 0 else 0
duration_formatted = str(timedelta(seconds=int(duration_seconds)))

print("="*70)
print("1. VIDEO INFORMATION")
print("="*70)
print(f"FPS: {fps}")
print(f"Total Frames: {total_frames}")
print(f"Frame Dimensions: {frame_width}x{frame_height}")
print(f"Duration: {duration_formatted} ({duration_seconds:.2f} seconds)")
print()
```

## ==================== 2. DISPLAY FIRST 10 FRAMES ON CONSOLE ====================

```python
print("="*70)
print("2. FIRST 10 FRAMES ANALYSIS")
print("="*70)

frame_count = 0
second = 1
frame_list = []

while True:
    isvalid, frame = video_process.read()
    
    if not isvalid:
        print("✗ Failed to capture frame")
        break
    
    gray_image_frame = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
    resized_frame = cv2.resize(gray_image_frame, (300, 200))
    
    # Display first 10 frames on console
    if frame_count < 10:
        mean_intensity = np.mean(resized_frame)
        std_intensity = np.std(resized_frame)
        print(f"Frame {frame_count + 1:3d} | Size: 300x200 | Mean Intensity: {mean_intensity:.2f} | Std Dev: {std_intensity:.2f}")
        frame_list.append(resized_frame)
    
    cv2.imshow('Video Frame', resized_frame)
    
    # Save every 10th frame with second counter
    if frame_count == 9:
        cv2.imwrite(f'frame_{frame_count + 1}_{second}.jpg', resized_frame)
        second += 1
        frame_count = 0
    else:
        frame_count += 1
    
    if cv2.waitKey(1) & 0xFF == ord('q'):
        print("\n✓ Video processing stopped by user")
        break

video_process.release()
cv2.destroyAllWindows()
```

## ==================== 3. TOTAL FRAMES SUMMARY ====================

```python
print()
print("="*70)
print("3. TOTAL FRAMES AND DURATION")
print("="*70)
print(f"Total Frames Processed: {total_frames}")
print(f"Video Duration: {duration_formatted}")
print(f"Frames Per Second (FPS): {fps}")
print()

```

## ==================== 4. BRIGHTNESS ANALYSIS OF FIRST 10 FRAMES ====================


```python
print("="*70)
print("4. BRIGHTNESS ANALYSIS - FIRST 10 FRAMES")
print("="*70)

brightness_values = []
for idx, frame in enumerate(frame_list):
    brightness = np.mean(frame)
    brightness_values.append(brightness)
    print(f"Frame {idx + 1:2d}: Brightness = {brightness:.2f} {'█' * int(brightness / 10)}")

avg_brightness = np.mean(brightness_values) if brightness_values else 0
max_brightness = np.max(brightness_values) if brightness_values else 0
min_brightness = np.min(brightness_values) if brightness_values else 0

print(f"\nBrightness Statistics:")
print(f"  Average: {avg_brightness:.2f}")
print(f"  Maximum: {max_brightness:.2f}")
print(f"  Minimum: {min_brightness:.2f}")
print()

```

## ==================== 5. EDGE DETECTION ON FIRST 10 FRAMES ====================

```python
print("="*70)
print("5. EDGE DETECTION - FIRST 10 FRAMES")
print("="*70)

edge_counts = []
for idx, frame in enumerate(frame_list):
    edges = cv2.Canny(frame, 50, 150)
    edge_pixel_count = np.sum(edges > 0)
    edge_counts.append(edge_pixel_count)
    percentage = (edge_pixel_count / (300 * 200)) * 100
    print(f"Frame {idx + 1:2d}: Edge Pixels = {edge_pixel_count:6d} ({percentage:5.2f}%) {'▓' * int(percentage / 5)}")

avg_edges = np.mean(edge_counts) if edge_counts else 0
print(f"\nAverage Edge Pixels: {avg_edges:.2f}")
print()

```

## ==================== 6. MOTION DETECTION BETWEEN CONSECUTIVE FRAMES ====================

```python
print("="*70)
print("6. MOTION DETECTION - CONSECUTIVE FRAMES")
print("="*70)

motion_values = []
for idx in range(len(frame_list) - 1):
    diff = cv2.absdiff(frame_list[idx], frame_list[idx + 1])
    motion = np.sum(diff) / (300 * 200)
    motion_values.append(motion)
    print(f"Frame {idx + 1} → {idx + 2}: Motion Score = {motion:.4f} {'▓' * int(motion * 50)}")

if motion_values:
    avg_motion = np.mean(motion_values)
    max_motion = np.max(motion_values)
    print(f"\nAverage Motion: {avg_motion:.4f}")
    print(f"Maximum Motion: {max_motion:.4f}")

print()
print("="*70)
print("✓ VIDEO PROCESSING COMPLETED SUCCESSFULLY!")
print("="*70)
```
