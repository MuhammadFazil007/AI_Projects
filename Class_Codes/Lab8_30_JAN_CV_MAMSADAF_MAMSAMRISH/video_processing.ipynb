### Video Processing

```python
import cv2

video_process = cv2.VideoCapture('Screen Recording 2025-12-17 022407.mp4')

if not video_process.isOpened():
    print("Unable To Open The Video")
    exit()

# Total frames count
total_frames = int(video_process.get(cv2.CAP_PROP_FRAME_COUNT))
print("Total Number of Frames in Video:", total_frames)

frame_number = 0

while True:
    isvalid, frame = video_process.read()

    if not isvalid:
        break

    # Convert to grayscale & resize
    gray_frame = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
    resized_frame = cv2.resize(gray_frame, (300, 200))

    # Save only first 10 frames
    if frame_number < 10:
        cv2.imwrite(f"frame_{frame_number+1}.jpg", resized_frame)

    # Show video
    cv2.imshow("Video Frame", resized_frame)

    frame_number += 1

    # Exit on 'q' key
    if cv2.waitKey(1) & 0xFF == ord('q'):
        print("Program stopped by user.")
        break

video_process.release()
cv2.destroyAllWindows()

print("First 10 frames saved successfully.")
```
