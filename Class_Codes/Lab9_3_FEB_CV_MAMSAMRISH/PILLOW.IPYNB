```python
from PIL import Image
import matplotlib.pyplot as plt
```

### Reading the Image

```python
img = Image.open("pexels-gabriele-brancati-32566116-7627322.jpg")
pixels = list(img.getdata())
min_intensity = min(pixels)
max_intensity = max(pixels)
print(f" Minimum Intensity Pixel Is {min_intensity}")
print(f" Maximum Intensity Pixel Is {max_intensity}")
```

### Converting to the Grayscale

```python
image_gray = img.convert("L")
```

### Resizing the Image to 500 x 500


```python
image_gray = image_gray.resize((500, 500))
```

### Finding Out The Maximum & Minimum Intensity Values

```python
pixels = list(image_gray.getdata())
min_intensity = min(pixels)
max_intensity = max(pixels)
print(f" Minimum Intensity Pixel Is {min_intensity}")
print(f" Maximum Intensity Pixel Is {max_intensity}")
```

### Gray Scale Histogram

```python
manual_hist = [0] * 256
for p in pixels:
    manual_hist[p] += 1

builtin_hist = image_gray.histogram()    
plt.figure()
plt.title("Manual Gray Scale Histogram")
plt.xlabel(" Pixel Intensity")
plt.ylabel(" Number of Pixels")
plt.plot(manual_hist, color='black')

plt.figure()
plt.title("Built In Gray Scale Histogram")
plt.xlabel(" Pixel Intensity")
plt.ylabel(" Number of Pixels")
plt.plot(builtin_hist, color='red')


```
