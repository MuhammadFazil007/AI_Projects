```python
import cv2
from PIL import Image


video = cv2.VideoCapture('Screen Recording 2025-12-17 022407.mp4')

if not video.isOpened():
    print("Unable to open video")
    exit()

# Total frames count
total_frames = int(video.get(cv2.CAP_PROP_FRAME_COUNT))
print("Total Frames:", total_frames)

frame_no = 0

while True:
    ret, frame = video.read()
    if not ret:
        break

    # BGR → RGB
    frame_rgb = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)

    # OpenCV → PIL
    pil_img = Image.fromarray(frame_rgb)

    # PIL processing
    gray = pil_img.convert("L")
    resized = gray.resize((300, 200))

    # Save only first 10 frames
    if frame_no / 10 :
        resized.save(f"frame_{frame_no+1}.jpg")

    frame_no += 1

    # Exit key (works in Jupyter + normal Python)
    if cv2.waitKey(1) & 0xFF == ord('q'):
        print("Stopped by user")
        break

video.release()
cv2.destroyAllWindows()

print("First 10 frames saved successfully.")
```
