### Important Libraries to Import

```python
import numpy
import cv2
import matplotlib.pyplot as plt
```

### Loading & Resizing the Image

```python
image = cv2.imread("pexels-alexgtacar-745150-1592384.jpg")
resized_image = cv2.resize(image, (500, 500))
resized_image2 = resized_image.copy()
```

### Gaussain Blur(Step 1)

```python
g_blur = cv2.GaussianBlur(resized_image,(5,5),1.4)
```

### Canny Edge Detector

```python
canny_edges = cv2.Canny(g_blur, 100,200,apertureSize = 3)
```

```python
plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
plt.imshow(resized_image2, cmap='gray')
plt.title('Original Image')
plt.axis('off')
plt.subplot(1, 2, 2)
plt.imshow(canny_edges, cmap='gray')
plt.title('Canny Edges')
plt.axis('off')
plt.tight_layout()
plt.show()
```

### Task: Try different Canny thresholds

```python
thresh_pairs = [(50, 150), (100, 200), (150, 250)]
titles = [f'Canny ({lo}, {hi})' for lo, hi in thresh_pairs]
results = [cv2.Canny(g_blur, lo, hi, apertureSize=3) for lo, hi in thresh_pairs]
 
plt.figure(figsize=(15, 5))
for i in range(3):
    plt.subplot(1, 3, i + 1)
    plt.imshow(results[i], cmap='gray')
    plt.title(titles[i])
    plt.axis('off')
plt.suptitle('Canny with Different Threshold Pairs')
plt.tight_layout()
plt.show()
```
